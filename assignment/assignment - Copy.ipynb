```python
import time
import requests
from bs4 import BeautifulSoup

rows = []
page = 1

while True:
    url = f"https://www.scrapethissite.com/pages/forms/?page_num={page}"
    print(f"Scraping page {page}: {url}")
    
    response = requests.get(url)
    soup = BeautifulSoup(response.text, 'html.parser')
    
    # 1. Extract data from the current page
    # (Assuming parse_and_extract_rows accepts a BeautifulSoup object or HTML text)
    page_rows = parse_and_extract_rows(response.text)
    rows.extend(page_rows)
    
    # 2. Check if the 'Next' (») button exists in the pagination
    next_button = soup.find('a', {'aria-label': 'Next'})
    
    if not next_button:
        print("No more pages found. Finishing scrap.")
        break
        
    # 3. Move to next page and pause to respect the server
    page += 1
    time.sleep(1)

print(f"Successfully scraped {len(rows)} rows across {page} pages.")
```
